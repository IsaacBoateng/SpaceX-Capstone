```python
import csv, sqlite3, prettytable
prettytable.DEFAULT = prettytable.PLAIN_COLUMNS
con = sqlite3.connect('my_data1.db')
cur = con.cursor()
%sql sqlite:///my_data1.db
```

```python
import pandas as pd
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv')
df.to_sql('SPACEXTBL', con, if_exists='replace', index=False, method='multi')
```

```python
%sql create table SPACEXTABLE as select * from SPACEXTBL where Date is not null
```

## Task 1: Display unique launch sites

```python
%sql SELECT DISTINCT Launch_Site FROM SPACEXTABLE
```

## Task 2: Launch sites begin with CCA

```python
%sql SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5
```

## Task 3: Total payload mass by NASA (CRS)

```python
%sql SELECT SUM(PAYLOAD_MASS__KG_) FROM SPACEXTABLE WHERE Customer = 'NASA (CRS)'
```

## Task 4: Average payload mass by F9 v1.1

```python
%sql SELECT AVG(PAYLOAD_MASS__KG_) FROM SPACEXTABLE WHERE Booster_Version LIKE 'F9 v1.1%'
```

## Task 5: First successful ground pad landing date

```python
%sql SELECT MIN(Date) FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (ground pad)'
```

## Task 6: Boosters with success on drone ship (4000-6000 kg)

```python
%sql SELECT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000
```

## Task 7: Total number of successful and failure mission outcomes

```python
%sql SELECT Mission_Outcome, COUNT(*) as Count FROM SPACEXTABLE GROUP BY Mission_Outcome
```

## Task 8: Boosters that carried maximum payload

```python
%sql SELECT Booster_Version FROM SPACEXTABLE WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE)
```

## Task 9: Failed drone ship landings in 2015

```python
%sql SELECT substr(Date,6,2) as Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE Landing_Outcome = 'Failure (drone ship)' AND substr(Date,0,5) = '2015'
```

## Task 10: Rank landing outcomes 2010-06-04 to 2017-03-20

```python
%sql SELECT Landing_Outcome, COUNT(*) as Count FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Count DESC
```
