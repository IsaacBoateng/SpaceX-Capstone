```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
df.head()
```

## Flight Number vs Launch Site

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=4)
plt.xlabel('Flight Number', fontsize=15)
plt.ylabel('Launch Site', fontsize=15)
plt.show()
```

## Payload Mass vs Launch Site

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=4)
plt.xlabel('Payload Mass (kg)', fontsize=15)
plt.ylabel('Launch Site', fontsize=15)
plt.show()
```

## Success Rate by Orbit

```python
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
sns.barplot(x='Orbit', y='Class', data=orbit_success)
plt.xlabel('Orbit', fontsize=15)
plt.ylabel('Success Rate', fontsize=15)
plt.title('Success Rate by Orbit Type')
plt.xticks(rotation=45)
plt.show()
```

## Flight Number vs Orbit

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=3)
plt.xlabel('Flight Number', fontsize=15)
plt.ylabel('Orbit', fontsize=15)
plt.show()
```

## Payload Mass vs Orbit

```python
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=3)
plt.xlabel('Payload Mass (kg)', fontsize=15)
plt.ylabel('Orbit', fontsize=15)
plt.show()
```

## Yearly Success Trend

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly = df.groupby('Year')['Class'].mean().reset_index()
plt.plot(yearly['Year'], yearly['Class'], marker='o')
plt.xlabel('Year')
plt.ylabel('Success Rate')
plt.title('Launch Success Yearly Trend')
plt.grid(True)
plt.show()
```
